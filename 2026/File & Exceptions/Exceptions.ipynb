### Handling the ZeroDivisionError Exception

```python
print(5/0)
```

```python
# Using try-except Blocks
try:
    print(5/0)
except ZeroDivisionError:
    print('You can not divide by zero!')
```

```python
# Any code that depends on the try block executing successfully goes in the else block

print("Give two numbers, I will devide them.")
print("Enter q to quit.")

while True:
    first_number = input("\nFirst Number: ")
    if first_number == 'q':
        break
    second_number = input("\nSecond Number: ")
    if second_number == 'q':
        break
    try:
        answer = int(first_number)/int(second_number)
    except ZeroDivisionError:
        print('You can not divide by zero!')
    else:
        print(answer)
```

### Handling the FileNotFoundError Exception

```python
filename = 'alice.txt'

with open(filename, encoding='utf-8') as file_content:
    contents = file_content.read()

print(contents)
```

```python
filename = 'alice.txt'

try:
    with open(filename, encoding='utf-8') as file_content:
        contents = file_content.read()
except FileNotFoundError:
    print(f"Sorry, the file {filename} does not exixst")
```

### Analysing Text

```python
title = 'Alice in Wonderland'
title.split()

'''
split() method separates a string into parts wherever it finds a 
space and stores all the parts of the string in a list.
'''
```

```python
filename = 'books.txt'

try:
    with open(filename, encoding='utf-8') as file_contents:
        contents = file_contents.read()
except FileNotFoundError:
    print(f'Sorry, the  file {filename} does not exitst')
else:
    words = contents.split()
    number_of_words = len(words)
    print(f"The file {filename} has about {number_of_words} words")
```

```python
def count_words(filename):
    try:
        with open(filename, encoding='utf-8') as file_contents:
            contents = file_contents.read()
    except FileNotFoundError:
        print(f'Sorry, the  file {filename} does not exitst')
    else:
        words = contents.split()
        number_of_words = len(words)
        print(f"The file {filename} has about {number_of_words} words")
        
filename = 'alice.txt'
count_words(filename)
```

```python
# multiple files
filenames = ['alice.txt', 'programming.txt', 'books.txt']
for filename in filenames:
    count_words(filename)
```
