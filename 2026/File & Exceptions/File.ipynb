### Reading an Entire File

```python
with open('pi_digit.txt') as file_object:
    contents = file_object.read()

print(contents)

print('Finished the file content')


with open('pi_digit.txt') as file_object:
    contents2 = file_object.read()

print(contents2.rstrip())

'''
To work any file, like writing and printing, first need to open the file.
open() needs a parameter, which is the name of the file.
'''

# read() uses for readig the entire contents of the files.
```

### File Path

```python
# Relative Path: Look for a given location
with open('text_files/first.txt') as file_object:
    contents = file_object.read()
print(contents.title())
```

### Reading Line by Line

```python
filename = 'pi_digit.txt'

with open(filename) as file_object:
    for line in file_object:
        print(line)

print("Finished")

with open(filename) as file_object:
    for line in file_object:
        print(line.rstrip())
```

### Making a List of Lines from a File

```python
# readlines() method takes each line from a file and stores in a list
with open(filename) as file_object:
    lines = file_object.readlines()

print(lines)
for line in lines:
    print(line)
    # print(line.rstrip())
```

```python
# wirkign with the file contents
pi_string = ''
for line in lines:
    pi_string += line.rstrip()

print(pi_string)
print(f"Length of the content is: {len(pi_string)}")
```

### Writing to an Empty file

```python
file_name = 'programming.txt'

with open(file_name, 'w') as file_object:
    file_object.write('I love programming')
    file_object.write('\nI like problem solving') # writing multiple line

```
