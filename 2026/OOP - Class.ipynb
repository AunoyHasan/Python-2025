### Creating a Class

```python
# Defining a class
class Dog:
    '''
    The __init__() method is a special method that Python runs 
    automatically whenever we create a new instance based on a class.
    '''
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    '''
    The self parameter is required in the method definition, and it must come 
    first before the other parameters. It must be included in the definition 
    because when Python calls this method later, the method call will 
    automatically pass the self argument.
    ''' 
        
    def sit(self):
        print(f'\n{self.name} is now sitting')
    
    def roll_over(self):
        print(f"\n{self.name} rolled over")

# class keyword is used to define a class.
# __init__ is a constructor (runs automatically when object is created).
# self refers to the current object.


# Making an instance from a class
my_dog = Dog('Willie', 6)

print(f"My dog's name is {my_dog.name}.") # Accessing Attributes
print(f"My dog is {my_dog.age} years old.") # Accessing Attributes

my_dog.sit()  # calling method

my_dog.roll_over()  # calling method
```

### Working with Classes and Instances

```python
class Pizza:
    def __init__(self, size, poppings):
        self.size = size
        self.poppings = poppings
    
    def __str__(self):
        return f"Pizza ({self.size} inches) with {', '.join(self.poppings)}"
            
my_pizza = Pizza(12, ['Chicken', 'Extra Cheese', 'Tomato', 'Beef'])
print(my_pizza)
```

### Setting a Default Value for an Attribute

```python
class Car:
    def __init__(self, manufacturer, model, year):
        self.manufacturer = manufacturer
        self.model = model
        self.year = year
        self.odometer_reading = 0 # default value for an attribure
        
    def get_descriptive_name(self):
        long_name = f"{self.year} {self.manufacturer} {self.model}"
        return long_name.title()
    
    def read_odometer(self):
        print(f"This car has {self.odometer_reading} miles on it.")
    
my_new_car = Car('audi', 'a4', 2019)
print(my_new_car.get_descriptive_name())
my_new_car.read_odometer()
```

### Modifying an Attribute Value

```python
class Cat:
    def __init__(self, size, color, type):
        self.size = size
        self.color = color
        self.type = type
        self.legs = 4 # Default attribute
        
    def cat_information(self):
        print(f"My cat's size is {self.size.title()}")
        print(f"My cat's color is {self.color.title()}")
        print(f"My cat's type is {self.type.title()}")
        print(f"My cat has {self.legs} legs")
    
my_cat = Cat('medium', 'white', 'desi')
my_cat.legs = 3
my_cat.cat_information()
```
