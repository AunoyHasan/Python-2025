### Defining a dictionary

```python
alien = {'color':'green', 'point':5}
print(alien)
```

```python
alien['color']
```

### Adding new value

```python
alien['goal'] = "Come in Earch"
alien['height'] = 5.7

print(alien)
```

### Modifying

```python
alien['point'] = 100
print(alien)
```

### Removing

```python
del alien['point']
print(alien)
```

### Looping through all key  value pair

```python
favourite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python'
}

for  name, language in favourite_languages.items():
    print(f"{name.title()}'s favourite language is {language.title()}.'")
```

```python
for name in favourite_languages.keys():
    print(name.title())
```

```python
for name in favourite_languages.values():
    print(name.title())
```
