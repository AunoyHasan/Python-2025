### Defining a Function

```python
# Defining a funciton
def greet_user():
    print("Welcome home")

# Calling a function
greet_user()
```

```python
# Passing Information to a Function
def greeting_user(username):
    print(f"Welcome home, {username}")

# Calling a funciton
greeting_user("Aunoy")
```

### Passing arguments

```python
# Passing positional arguments
def describe_pet(animal_type, animal_name):
    print(f"I have a {animal_type}")
    print(f"My {animal_type}'s name is {animal_name}")

describe_pet('Dog', 'Tommy')
```

```python
# Keyword Arguments
describe_pet(animal_type = "Cat", animal_name = "Jessy")
describe_pet(animal_name = "Tara", animal_type = "Horse")
```

### Default Value

```python
def my_pet(animal_name, animal_type = "Dog"):
    print(f"I have a {animal_type}")
    print(f"My {animal_type}'s name is {animal_name}'")
    
my_pet(animal_name = "Donkey")
my_pet("Lion", "terabox")
```

### Return Value

```python
def full_name(first_name, last_name):
    full_name = first_name.title() + " " +last_name.title()
    return full_name

    # full_name = f"{first_name} {last_name}"
    # return full_name.title()

name = full_name("Mahmudul Hasan", "Aunoy")
print("My full name is ", name)
```

### Optional Argument

```python
def get_formatted_name(first_name, last_name, middle_name = ''):
    if middle_name:
        full_name = f"{first_name} {middle_name} {last_name}"
    else:
        full_name = f"{first_name} {last_name}"
    
    return full_name

name = get_formatted_name("Albert", "Einstein")
middle_name = get_formatted_name("Hossain", "Ershad", "Mohammed")

print(name)
print(middle_name)
```

### Returning a dictionary

```python
def build_person(first_name, last_name):
    person = {'first': first_name, 'last': last_name}
    return person

musician = build_person('Kishore', 'Kumar')
print(musician)
```

### Passing a list

```python
def names_list(names):
    for  name in names:
        print(f"Hello {name.title()}")
        
usernames = ['james', 'kishore', 'aunoy', 'abid', 'mahedi']
names_list(usernames)
```

### Modifying a list in a function

```python
def processing_order(queue_foods, finished_foods):
    order_number = 1
    while queue_foods:
        food_name = queue_foods.pop()
        print(f'Order No {order_number} and food name {food_name}')
        finished_foods.append(food_name.title())
        order_number = order_number + 1

def handed_foods(finished_foods):
    for finish_food in finished_foods:
        print('Handed', finish_food.title())

queue_foods = ['berger', 'hamberger', 'hotdog', 'durum', 'coca-cola'] 
finished_foods = []

processing_order(queue_foods, finished_foods)
handed_foods(finished_foods)
```

### Passing an Arbitrary Number of Arguments

```python
def make_pizza(*toppings):
    print(toppings)

make_pizza('pepperoni')
make_pizza('pepperoni', 'mushroom', 'chicken', 'cheese', 'green peppers')
```

```python
def make_burger(*toppings):
    print("\nMaking a burger with the following toppings: ")
    for topping in toppings:
        print(f"- {topping.title()}")
        
make_burger('Beaf')
make_burger('Chicken', 'Honey', 'quequmber', 'tomator', 'chesse')
```

### Mixing positional and arbitary arguments

```python
def make_biriyani(quantity, *toppings):
    print(f"\nMaking {quantity} kg biriani with the following topings: ")
    for topping in toppings:
        print(f"- {topping.title()}")

make_biriyani(1, 'chicken', 'potato')
make_biriyani(3, 'chicken', 'potato', 'beaf', 'spice')
```

### Using Arbitrary Keyword Arguments

```python
#  key-value pairs arguments

def build_profile(first, last, **user_info):
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

user_profile = build_profile('albert', 'einstein', locaiton = 'princeton', field = 'quantam physics')
print(user_profile)
```
