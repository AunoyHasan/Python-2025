```python
# Parent class Car
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
        
    def get_descriptive_name(self):
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title();
    
    def read_odometer(self):
        print(f"This car has {self.odometer_reading} miles on it.")
        
    def update_odometer(self, milage):
        if milage >= self.odometer_reading:
            self.outdoor_reading = milage
        else:
            print("You can not roll back an adometer!")
    
    def increment_odometer(self, miles):
        self.odometer_reading += miles
        

# the child class: ElectricCar

class ElectricCar(Car):
    def __init__(self, make, model, year):
        # Initialize atttribures of the parent class
        super().__init__(make, model, year)
        self.battery_size = 75
        # super() function is a special function that allows you to call a method from the parent class
        
    def describe_battery(self):
        print(f"This car has a {self.battery_size}-kwh battery")

# creating object of the child class
my_tesla = ElectricCar('tesla', ' model s', 2019)
print(my_tesla.get_descriptive_name())
my_tesla.describe_battery()
```

### Instances as Attributes

```python
class Battery:
    def __init__(self, battery_size = 125):
        self.battery_size = battery_size
    
    def describe_battery(self):
        print(f"This car has a {self.battery_size}-kwh battery.")
        
        
class NonElectricCar(Car):
    def __init__(self, make, model, year):
        # Initialize atttribures of the parent class
        super().__init__(make, model, year)
        self.battery = Battery()
        # super() function is a special function that allows you to call a method from the parent class
        

# creating object of the child class
my_toyota = NonElectricCar('Toyota', 'nissan 2', 2001)
print(my_toyota.get_descriptive_name())
my_toyota.battery.describe_battery()
```
