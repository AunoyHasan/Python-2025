# List

### Defining a list and printng

```python
fruits = ['banana', 'orange', 'apple', 'pineapple', 'watermalone', 'guava']
```

```python
# print the list
fruits
```

```python
# print the list
print(fruits)
```

```python
# print using for loop
for fruit in  fruits:
    print(fruit)
```

```python
# print using while loop
i = 0;
while i<len(fruits):
    print(fruits[i])
    i = i+1
```

### Accessing the list

```python
# pirnt the first element
print(fruits[0])
```

```python
# print the last element
print(fruits[-1])
```

### Adding in the list

```python
# using append fucntion
fruits.append('Mango');
print(fruits)
```

```python
# using insert function
fruits.insert(3, 'Grapes')
fruits.insert(4, 'Strawberry')
print(fruits)
```

### Removing element form the list

```python
# using del keyword
del fruits[4]
print(fruits)
```

```python
# using pop function - the last element will be deleted. 
fruits.pop()
print(fruits)
```

```python
# specific index using pop funciton
fruits.pop(3);
print(fruits)
```

```python
# removing an item by value
fruits.remove('pineapple')
fruits.remove('watermalone')
print(fruits)
```

### Sorting a list

```python
# Ascending sort
fruits.sort()
print(fruits)
```

```python
# Descending sort
fruits.sort(reverse=True)
print(fruits)
```

```python
# sorting a list tempararily with sorted function
cars = ["Toyota", "BMW", "Tesla", "Ford", "Audi"]
print(f"Main list of cars {cars}")
print(f"Sorted list of cars {sorted(cars)}")
print(f"After sorting the list of cars {cars}")
```

```python
# reverse order
cars.reverse()
print(cars)
```

### List in input function

```python
# Mehtod 1
animals = input("Enter animals names separated by space: ").split()

print("Your list:", animals)
```

```python
# Method 2
social_media = input("Enter social media names separated by commas: ").split(",")

print("Your list:", social_media)
```

```python
# Method 3: using loop
companies = []

n = int(input('How many companies you want to add? '))

for i in range(n):
    company = input(f"Enter company name {i+1}: ")
    companies.append(company)
    
print('Company list is ', companies)
```

### Conditional Statements in List

```python
# if statement
print("Players list: ", players[:])
for player in players:
    if player == 'abid':
        print(player.upper())
    else:
        print(player.title())
```

```python
# Checking a values in a list or not
print(f"Players list {players[:]}")

serarching_player = str(input("Enter the player name you want to search: "))
find_value = 0;

for player in players:
    if serarching_player == player:
        find_value = 1;

if find_value == 1:
    print(f"Yes, {serarching_player} is available in list")
else:
    print(f"No, {serarching_player} is available in the list")
```

```python
# using multiple list
available_countries = ['Bangladesh', 'Japan', 'China', 'Denmark', 'Germany', 'Finland', 'Korea']
requested_countries = ['Bangladesh', 'Japan', 'China', 'Thailand', 'South Africa']

for requested_country in requested_countries:
    if requested_country in available_countries:
        print(f"Adding {requested_country}")
    else:
        print(f"Sorry, we do not have {requested_country}")
```
